```python
import pandas as pd
from sqlalchemy import create_engine

# 1. Database Connection URL (သင့်ရဲ့ Neon.tech Database URL ကို ထည့်ပါ)
# 'postgresql://' အစား 'postgresql+psycopg2://' ကို အသုံးပြုရပါမည်[cite: 2]
DATABASE_URL = "postgresql+psycopg2://neondb_owner:npg_IRB5UjTfCL9o@ep-snowy-wildflower-a14gbmyt-pooler.ap-southeast-1.aws.neon.tech/neondb?sslmode=require"

# 2. SQLAlchemy Engine ဖန်တီးခြင်း[cite: 2]
engine = create_engine(DATABASE_URL)
print("Database connected successfully!")

# 3. SQL Query ဖြင့် Customer Segments များကို ဆွဲထုတ်ခြင်း
query = """
SELECT * FROM customer_segments;
"""

# 4. Pandas DataFrame သို့ ပြောင်းလဲခြင်း
df = pd.read_sql(query, con=engine)

# 5. Data ပုံစံအချို့ကို စစ်ဆေးခြင်း
print(f"Total Rows Fetched: {len(df)}")
display(df.head())

# 6. Tableau အတွက် CSV ဖိုင်အဖြစ် Save လုပ်ခြင်း
df.to_csv("customer_segments.csv", index=False)
print("customer_segments.csv file generated successfully!")
```
